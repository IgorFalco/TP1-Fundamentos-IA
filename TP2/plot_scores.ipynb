```python

import numpy as np
import matplotlib.pyplot as plt

# Load the episode scores
scores = np.load("episode_scores.npy")

# Plotting
plt.figure(figsize=(12, 6))
plt.plot(scores)
plt.title("Pontuação por Episódio durante o Treinamento de Marvin")
plt.xlabel("Episódio")
plt.ylabel("Pontuação Total")
plt.grid(True)
plt.savefig("episode_scores_plot.png")
plt.close()

print("Plot saved as episode_scores_plot.png")


```
